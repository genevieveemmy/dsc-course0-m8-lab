#### construct a derived column tracking whether an aircraft was destroyed or not.
* **DESTROYED** = Hull Total Loss (Write-Off): The aircraft is beyond repair or completely consumed. The insurer pays out the full insured value (100% hull payout). This is the exact definition of "total destruction."

* **SUBSTANTIAL** = Hull Repair Claim: The aircraft suffered major structural damage (e.g., bent wing spars, crushed landing gear), but the airframe is repairable. The insurer pays for repairs/overhaul, which is usually a fraction of the total hull value.

I will isolate `Total Destruction` (Hull Loss) as events coded `DESTROYED` because these represent 100% total loss payouts for underwriters. `SUBSTANTIAL` damage events are analyzed separately as repair claims, ensuring our total loss risk model is not diluted by repairable airframe incidents

```python
# 1. Create binary column (1 if DESTROYED, 0 for anything else)
filtered_df["Is_Destroyed"] = (
    filtered_df["Aircraft.damage"].astype(str).str.strip().str.upper() == "DESTROYED"
).astype(int)

# 2. Check value counts to confirm
print(filtered_df["Is_Destroyed"].value_counts())
```

### Data Cleaning Tasks: Professional & Active Aircraft Filter

| Cleaning Task | Operations Required | Rationale & Client Constraint |
| :--- | :--- | :--- |
| **1. Filter Professional Builds** | Filter `Amateur.Built == 'NO'` | Restricts scope to professionally manufactured aircraft, removing experimental kit builds. |
| **2. Standardize Manufacturer Names (`Make`)** | Trim whitespace, uppercase, and apply regex mapping | Consolidates spelling variations (e.g., `CESSNA AIRCRAFT CO`, `CESSNA CO`, `CESSNA`) into single canonical make names. |
| **3. Apply Statistical Threshold ($N \ge 50$)** | Filter `Make` counts where total incident volume $N \ge 50$ | Removes low-frequency manufacturers to prevent small-sample noise from skewing risk rates. |

#### Step 1: Filter to remove non professional builds
 * Our client is only interested in airplane makes/models that are professional builds and could potentially still be active.
 * An amateur-built aircraft—often called a homebuilt aircraft—is a plane that is constructed by an individual or group of individuals rather than built in a professional factory.
 * Lets explore `Amateur.Build column`

```python
filtered_df['Amateur.Built'].value_counts()
```

```python
#fill nulls with unknown
filtered_df['Amateur.Built'] = filtered_df['Amateur.Built'].fillna("Unknown")
```

```python
# filtered_df['Amateur.Built'].value_counts() -> we have only 101 unknowns, its safe to drop them
#we take only professional builds represented by NO
filtered_df = filtered_df[
    filtered_df["Amateur.Built"].astype(str).str.strip().str.upper() == "NO"
]
```

#### Step2: Standardize name of 'Make'


```python
#standardize make names 
filtered_df["Make"] = (
    filtered_df["Make"].fillna("UNKNOWN").astype(str).str.strip().str.upper()
)
filtered_df["Make"]
```

```python
# Mapping dictionary for duplicate/inconsistent manufacturer names
make_mapping = {
    # Airbus
    "AIRBUS INDUSTRIE": "AIRBUS",
    # Bombardier
    "BOMBARDIER INC": "BOMBARDIER",
    # Cirrus
    "CIRRUS DESIGN CORP.": "CIRRUS",
    "CIRRUS DESIGN CORP": "CIRRUS",
    # Robinson
    "ROBINSON HELICOPTER": "ROBINSON",
    "ROBINSON HELICOPTER COMPANY": "ROBINSON",
    # De Havilland
    "DEHAVILLAND": "DE HAVILLAND",
    # Grumman variants
    "GRUMMAN AMERICAN": "GRUMMAN",
    "GRUMMAN AMERICAN AVN. CORP.": "GRUMMAN",
    "GRUMMAN-SCHWEIZER": "GRUMMAN",
    "GRUMMAN ACFT ENG COR-SCHWEIZER": "GRUMMAN",
    # Ercoupe
    "ERCOUPE (ENG & RESEARCH CORP.)": "ERCOUPE",
    # Air Tractor
    "AIR TRACTOR INC": "AIR TRACTOR",
    # Aviat
    "AVIAT AIRCRAFT INC": "AVIAT",
    # Diamond
    "DIAMOND AIRCRAFT IND INC": "DIAMOND",
    # Rockwell
    "ROCKWELL INTERNATIONAL": "ROCKWELL",
    # Learjet
    "GATES LEARJET": "LEARJET",
    # Aerostar
    "SMITH, TED AEROSTAR": "AEROSTAR",
    # Flight Design
    "FLIGHT DESIGN GMBH": "FLIGHT DESIGN",
    # American Champion
    "AMERICAN CHAMPION AIRCRAFT": "AMERICAN CHAMPION",
    #BOEING
    "BOEING STEARMAN": "BOEING",
    #CHAMPION
    "AMERICAN CHAMPION": "CHAMPION",
}

# 1. Strip leading/trailing whitespace and convert to uppercase
filtered_df["Make"] = filtered_df["Make"].astype(str).str.strip().str.upper()

# 2. Replace variants with standard names
filtered_df["Make"] = filtered_df["Make"].replace(make_mapping)
```

#### Step4:Filter Make counts where total incident volume is >= 50


```python
# 1. Count incidents per manufacturer
make_counts = filtered_df["Make"].value_counts()

# 2. Retain only Makes with at least 50 incidents
robust_makes = make_counts[make_counts >= 50].index
filtered_df = filtered_df[filtered_df["Make"].isin(robust_makes)].copy()

# 3. Check remaining rows and qualifying Makes count
print(f"Remaining records: {len(filtered_df):,}")
print(f"Qualifying Makes: {filtered_df['Make'].nunique()}")
print(filtered_df["Make"].value_counts().head(10))
```

#### 1. cleaning Engine Type
* View counts of different values
* Fill nulls
* Standardize values

```python
filtered_df['Engine.Type'].value_counts()
```

```python
def clean_column_engine_type(df):
    df["Engine.Type"] = (
        df["Engine.Type"].fillna("UNKNOWN").astype(str).str.strip().str.upper()
    )
    unknowns = df["Engine.Type"].isin(["UNK", "NONE", "LR", ""])
    df.loc[unknowns, "Engine.Type"] = "UNKNOWN"

    return df



filtered_df = clean_column_engine_type(filtered_df)
filtered_df['Engine.Type'].value_counts()  
```

#### 2. Cleaning Weather column
* View counts of different values
* Fill nulls
* Standardize values

```python
filtered_df['Weather.Condition'].value_counts()
```

```python
# 1. Convert to uppercase (merges 'Unk' into 'UNK')
filtered_df["Weather.Condition"] = (
    filtered_df["Weather.Condition"]
    .fillna("UNKNOWN")
    .astype(str)
    .str.strip()
    .str.upper()
)

# 2. Replace 'UNK' with 'UNKNOWN'
filtered_df["Weather.Condition"] = filtered_df["Weather.Condition"].replace(
    "UNK", "UNKNOWN"
)

# Verify updated counts
print(filtered_df["Weather.Condition"].value_counts())
```

#### 3. Cleaning Number of Engines column
* View value counts
* Impute NaNs


```python
filtered_df['Number.of.Engines'].value_counts()
```

Since planes need at least one engine to fly , I will convert the 0.0 to NaN then impute with the median which was 1.00 in our `.describe`

```python
#convert 0.0 to NaN
filtered_df['Number.of.Engines'] = filtered_df['Number.of.Engines'].replace(0.0, np.nan)
# 2. Impute NaNs with median 
filtered_df["Number.of.Engines"] = filtered_df["Number.of.Engines"].fillna(filtered_df["Number.of.Engines"].median())
filtered_df['Number.of.Engines'].value_counts()
```

#### 4. Cleaning Purpose of flight column
* Impute nulls
* Standardize categories

```python
filtered_df['Purpose.of.flight'].value_counts()
```

```python
# 1. Uppercase and strip whitespace
filtered_df["Purpose.of.flight"] = (
    filtered_df["Purpose.of.flight"]
    .fillna("UNKNOWN")
    .astype(str)
    .str.strip()
    .str.upper()
)

# 2. Merge duplicate/public categories
filtered_df["Purpose.of.flight"] = filtered_df["Purpose.of.flight"].replace(
    {
        "AIR RACE SHOW": "AIR RACE / SHOW",
        "AIR RACE/SHOW": "AIR RACE / SHOW",
        "ASHO": "AIR RACE / SHOW",
        "PUBLIC AIRCRAFT - FEDERAL": "PUBLIC AIRCRAFT",
        "PUBLIC AIRCRAFT - STATE": "PUBLIC AIRCRAFT",
        "PUBLIC AIRCRAFT - LOCAL": "PUBLIC AIRCRAFT",
        "PUBS": "PUBLIC AIRCRAFT",
        "PUBL": "PUBLIC AIRCRAFT",
    }
)

# Verify clean counts
print(filtered_df["Purpose.of.flight"].value_counts())
```
